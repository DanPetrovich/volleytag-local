```
from ultralytics import YOLO
model = YOLO('yolo11n.pt')
result = model.train(
    data='/content/danya-detector-dataset/dataset.yaml',
    epochs=50, imgsz=960, batch=16, device=0, workers=2,
    patience=12, close_mosaic=8, cache='disk', seed=51,
    project='/content/runs', name='danya-detector-v1', plots=True
)
```
